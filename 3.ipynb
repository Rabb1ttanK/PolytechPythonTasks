```python
def memorize(func):
    cache = dict()
    def inner(*args, **kwargs):
        key = args + tuple(sorted(kwargs))
        if key not in cache:
            cache[key] = func(*args, **kwargs)
        return cache[key]
    return inner
```

```python
def timer(func=None, *, N=1): 

    # со скобками
    if func is None:
        def decorator(func):
            return timer(func, N=N)
        return decorator
    
    # без скобок
    @functools.wraps(func)
    def inner(*args, **kwargs):
        total_time = 0
        for irun in range(N):
            start_time = time.time()
            result = func(*args, **kwargs)
            end_time = time.time()

            delta_time = end_time - start_time
            total_time += delta_time
            print("Врема ", irun, " запуска составило ", delta_time)

        print(f"Среднее время выполнения функции: {total_time / N} секунд")
        return result
    
    return inner
```

```python
import time
import functools

def timer(N=2):
    def inner(func):

        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            total_time = 0
            for irun in range(N):
                start_time = time.time()
                result = func(*args, **kwargs)
                end_time = time.time()

                delta_time = end_time - start_time
                total_time += delta_time
                print("Врема ", irun, " запуска составило ", delta_time)

            print(f"Среднее время выполнения функции: {total_time / N} секунд")
            return result

        return wrapper
    return inner


@timer
def cycle(n):
    num = 0
    for i in range(n):
        num += i
    return num
 
cycle(5000000)


@timer(3)
def cycle2(n):
    """DOC!"""
    num = 0
    for i in range(n):
        num += i * i * i
    return num


cycle2(5000000)
```

```python
def once(func):
    def inner(*args, **kwargs):
        if not inner.has_run:
            inner.has_run = True
            return func(*args, **kwargs)
    inner.has_run = False
    return inner   
 
@once
def my_function(foo, bar):
    return foo+bar
```

```python
api_tokens = ["TRUE_TOKEN"]
def validate_api_token(validation_func):
    def decorator(func):
        @wraps(func)
        def decorated_function(*args, **kws):
            api_token = request.headers.get('Token')
            is_valid_api_token = validation_func(api_token)
            if is_valid_api_token:
                return func(*args, **kws)
            return 'Invalid API Token', 401
        return decorated_function
    return decorator


def simple_api_token_validation(api_token):
    return api_token in api_tokens
```

```python
from flask import Flask, request
from functools import wraps

app = Flask(__name__)


@app.route('/', methods=['GET'])
@validate_api_token(simple_api_token_validation)
def post_hello():
    return 'GET Hello, World!'


@app.route('/dima', methods=['GET'])
@validate_api_token(simple_api_token_validation)
def dima():
    return 'DIMA!'



app.run()
```
